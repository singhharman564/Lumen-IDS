```python
import pandas as pd
from pathlib import Path

LOGS = Path("../data/captures")

scan = pd.read_json(
    LOGS / "zeek_portscan_01" / "conn.log",
    lines=True
)

benign = pd.read_json(
    LOGS / "zeek_benign_01" / "conn.log",
    lines=True
)

print("Scan shape:", scan.shape)
print("Benign shape:", benign.shape)
print("Columns:", scan.columns.tolist())
```

```python
def summarize_traffic(df, name):
    print(f"\n--- {name} ---")
    print("Total flows:", len(df))
    print("Distinct destination ports:", df["id.resp_p"].nunique())
    print("\nConnection states:")
    print(df["conn_state"].value_counts(dropna=False))


summarize_traffic(benign, "Benign traffic")
summarize_traffic(scan, "Port scan traffic")
```

```python
print("Most common destination ports in scan traffic:")
print(scan["id.resp_p"].value_counts().head(10))

print("\nMost common destination ports in benign traffic:")
print(benign["id.resp_p"].value_counts().head(10))
```
